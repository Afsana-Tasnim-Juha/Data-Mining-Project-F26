```python
import pandas as pd

df = pd.read_csv(
    "data/raw/ev_full.csv",
    low_memory=False
)
print("Full dataset shape:", df.shape)

sample = df.sample(n=50000, random_state=42)
print("Sample shape:", sample.shape)


sample.to_csv("data/raw/ev_sample_50k.csv", index=False)
print("Saved: data/raw/ev_sample_50k.csv")
```
