```python
import pandas as pd

df = pd.read_csv("../data/raw/ev_sample_50k.csv")

print("Shape:", df.shape)
print("\nColumn data types:\n", df.dtypes)
print("\nMissing values per column:\n", df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print("Rows with Electric Range = 0:", (df["Electric Range"] == 0).sum())
print("\nState value counts:\n", df["State"].value_counts())
print("\nBase MSRP column exists:", "Base MSRP" in df.columns)
```

```python
print("Duplicate VINs:", df["VIN (1-10)"].duplicated().sum())
print("\nCAFV Eligibility values:\n", df["Clean Alternative Fuel Vehicle (CAFV) Eligibility"].value_counts())
```
